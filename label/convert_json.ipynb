## Convert to usable format

```python
import json
import re
from collections import defaultdict

# Load the original file
with open("final.json", "r") as f:
    data = json.load(f)

# Create a mapping from issue number to labels
issue_map = defaultdict(set)

for label, urls in data["categories"].items():
    for url in urls:
        match = re.search(r'/issues/(\d+)$', url)
        if match:
            number = int(match.group(1))
            issue_map[number].add(label)

# Convert to list of dicts in the desired format
output = [
    {"number": number, "labels": sorted(labels)}
    for number, labels in issue_map.items()
]

# Save to new JSON file
with open("issues-to-label.json", "w") as f:
    json.dump(output, f, indent=2)

print(f"Converted {len(output)} issues to issues-to-label.json")
```
